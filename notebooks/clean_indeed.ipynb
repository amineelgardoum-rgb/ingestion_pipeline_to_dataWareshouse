```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\lenovo\DataWarehouseProject\data\jobs_casablanca_indeed.tsv",
    sep="\t",
    dtype=str,
    keep_default_na=False,
    encoding="utf-8-sig"
)

# Strip hidden characters from all values
def clean_text(val):
    if isinstance(val, str):
        val = val.replace('\r', ' ')
        val = val.replace('\n', ' ')
        val = val.replace('\t', ' ')
        val = ' '.join(val.split())
    return val

df = df.apply(lambda col: col.map(clean_text))

# Add missing columns
df["work_from_home_type"] = ""
df["home_type"] = ""

df.to_csv(
    r"C:\Users\lenovo\DataWarehouseProject\data\jobs_casablanca_indeed_clean.tsv",
    sep="\t",
    index=False,
    encoding="utf-8",
    lineterminator="\n"
)

print(f"Done — {len(df)} rows, {len(df.columns)} columns saved.")
```
