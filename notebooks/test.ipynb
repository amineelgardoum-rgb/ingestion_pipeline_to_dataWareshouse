```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\lenovo\DataWarehouseProject\data\jobs_rekrute.tsv",
    sep="\t",
    dtype=str,
    keep_default_na=False,
    nrows=3  # just first 3 rows
)

# Check column count
print(f"Columns: {len(df.columns)}")
print(f"Column names: {df.columns.tolist()}")

# Check last column for hidden characters
last_col = df.columns[-1]
print(f"\nLast column name repr: {repr(last_col)}")
print(f"First value repr: {repr(df[last_col].iloc[0])}")

# Check first column for BOM
first_col = df.columns[0]
print(f"\nFirst column name repr: {repr(first_col)}")
```
